## 4. Konvertere eksterne data til slektstre

Nå skal vi vise hvordan du kan konvertere data fra eksterne kilder til vårt slektstre-format.


````python
# Konverter FamilySearch data til vårt format
def convert_familysearch_to_slektstre(familysearch_data):
    """
    Konverter FamilySearch data til vårt slektstre-format.
    """
    personer = []
    ekteskap = []
    
    # Konverter personer
    for fs_person in familysearch_data['persons']:
        # Parse navn
        navn_deler = fs_person['displayName'].split(' ')
        fornavn = navn_deler[0]
        etternavn = navn_deler[-1] if len(navn_deler) > 1 else ''
        
        # Parse datoer
        fødselsdato = None
        dødsdato = None
        try:
            if fs_person['birthDate']:
                fødselsdato = date.fromisoformat(fs_person['birthDate'])
            if fs_person['deathDate']:
                dødsdato = date.fromisoformat(fs_person['deathDate'])
        except:
            pass
        
        # Bestem kjønn
        kjønn = Gender.MALE if fs_person['gender'] == 'Male' else Gender.FEMALE
        
        # Opprett Person objekt
        person = Person(
            id=fs_person['id'],
            fornavn=fornavn,
            etternavn=etternavn,
            fødselsdato=fødselsdato,
            dødsdato=dødsdato,
            fødested=fs_person.get('birthPlace', ''),
            dødssted=fs_person.get('deathPlace', ''),
            kjønn=kjønn,
            notater=f"Importert fra FamilySearch (ID: {fs_person['id']})"
        )
        personer.append(person)
    
    # Konverter ekteskap (forenklet)
    for fs_person in familysearch_data['persons']:
        if fs_person['spouses']:
            for spouse_id in fs_person['spouses']:
                # Sjekk om ekteskapet allerede eksisterer
                eksisterer = any(
                    (e.partner1_id == fs_person['id'] and e.partner2_id == spouse_id) or
                    (e.partner1_id == spouse_id and e.partner2_id == fs_person['id'])
                    for e in ekteskap
                )
                
                if not eksisterer:
                    ekteskap_obj = Ekteskap(
                        id=f"e_{fs_person['id']}_{spouse_id}",
                        partner1_id=fs_person['id'],
                        partner2_id=spouse_id,
                        notater="Importert fra FamilySearch"
                    )
                    ekteskap.append(ekteskap_obj)
    
    return FamilieData(personer=personer, ekteskap=ekteskap)

# Konverter dataene
konvertert_data = convert_familysearch_to_slektstre(familysearch_data)

print("🔄 Konverterte FamilySearch data til slektstre-format:")
print(f"Personer: {len(konvertert_data.personer)}")
print(f"Ekteskap: {len(konvertert_data.ekteskap)}")

# Vis første person
if konvertert_data.personer:
    første_person = konvertert_data.personer[0]
    print(f"\n👤 Eksempel person: {første_person.fullt_navn}")
    print(f"   ID: {første_person.id}")
    print(f"   Født: {første_person.fødselsdato}")
    print(f"   Død: {første_person.dødsdato}")
    print(f"   Kjønn: {første_person.kjønn}")
    print(f"   Notater: {første_person.notater}")
````

````python
# Vis naboskap-matrisen for Helena-slektstreet
if 'helena_slektstre' in globals():
    import networkx as nx
    import pandas as pd
    import numpy as np
    
    print("📊 Naboskap-matrise for Helena-slektstreet")
    print("=" * 70)
    
    # Hent node-navnene (person-IDer)
    nodes = list(helena_slektstre.graph.nodes())
    n = len(nodes)
    
    # Opprett adjacency matrix manuelt (uten scipy)
    adj_matrix = np.zeros((n, n), dtype=int)
    
    # Fyll inn matrisen basert på kanter i grafen
    node_to_index = {node: i for i, node in enumerate(nodes)}
    for u, v in helena_slektstre.graph.edges():
        i = node_to_index[u]
        j = node_to_index[v]
        adj_matrix[i][j] = 1
        adj_matrix[j][i] = 1  # Urettet graf
    
    # Opprett en pandas DataFrame for pen formattering
    df_adjacency = pd.DataFrame(
        adj_matrix,
        index=nodes,
        columns=nodes
    )
    
    # Erstatt person-IDer med navn hvis mulig
    person_names = {}
    for person in helena_data.personer:
        person_names[person.id] = person.fullt_navn
    
    # Bytt ut IDer med navn
    df_adjacency.index = [person_names.get(node, node) for node in nodes]
    df_adjacency.columns = [person_names.get(node, node) for node in nodes]
    
    print(f"\nDimensjoner: {df_adjacency.shape[0]} x {df_adjacency.shape[1]}")
    print(f"Antall kanter (forbindelser): {helena_slektstre.graph.number_of_edges()}")
    print(f"\nMatrisen viser forbindelser mellom personer:")
    print("  1 = Det er en relasjon mellom personene")
    print("  0 = Ingen direkte relasjon\n")
    
    # Vis matrisen
    print(df_adjacency.to_string())
    
    # Beregn noen interessante statistikker
    print(f"\n📈 Statistikk:")
    degrees = dict(helena_slektstre.graph.degree())
    avg_degree = np.mean(list(degrees.values()))
    max_degree = max(degrees.values())
    max_degree_person = max(degrees, key=degrees.get)
    
    print(f"   Gjennomsnittlig antall forbindelser per person: {avg_degree:.2f}")
    print(f"   Person med flest forbindelser: {person_names.get(max_degree_person, max_degree_person)} ({max_degree} forbindelser)")
    
    # Vis hvem som har flest forbindelser
    print(f"\n👥 Top 3 personer med flest forbindelser:")
    sorted_degrees = sorted(degrees.items(), key=lambda x: x[1], reverse=True)
    for i, (person_id, degree) in enumerate(sorted_degrees[:3], 1):
        print(f"   {i}. {person_names.get(person_id, person_id)}: {degree} forbindelser")
    
else:
    print("⚠️ helena_slektstre er ikke lastet inn ennå.")
    print("   Kjør GEDCOM-import cellen først.")
````

### Undersøk manglende personer i grafen

La oss finne ut hvorfor noen ID-er finnes i grafen men ikke som personer.


````python
# Undersøk diskrepans mellom graf-noder og Person-objekter
if 'helena_slektstre' in globals() and 'helena_data' in globals():
    print("🔍 Analyserer diskrepans mellom graf og person-data")
    print("=" * 70)
    
    # Hent alle noder fra grafen
    graf_noder = set(helena_slektstre.graph.nodes())
    
    # Hent alle person-IDer fra helena_data
    person_ids = set(person.id for person in helena_data.personer)
    
    # Finn noder som er i grafen men ikke i person-data
    manglende_personer = graf_noder - person_ids
    
    # Finn personer som er i person-data men ikke i grafen
    ikke_i_graf = person_ids - graf_noder
    
    print(f"\n📊 Statistikk:")
    print(f"   Noder i grafen: {len(graf_noder)}")
    print(f"   Personer i helena_data: {len(person_ids)}")
    print(f"   Manglende personer (i graf, ikke i data): {len(manglende_personer)}")
    print(f"   Isolerte personer (i data, ikke i graf): {len(ikke_i_graf)}")
    
    if manglende_personer:
        print(f"\n⚠️ Noder i grafen som mangler Person-objekt:")
        for node_id in manglende_personer:
            print(f"   - {node_id}")
            
            # Finn hvem som refererer til denne personen
            neighbors = list(helena_slektstre.graph.neighbors(node_id))
            if neighbors:
                print(f"     Forbindelser til:")
                for neighbor_id in neighbors:
                    # Finn person-objektet
                    neighbor_person = next((p for p in helena_data.personer if p.id == neighbor_id), None)
                    if neighbor_person:
                        print(f"       → {neighbor_person.fullt_navn} ({neighbor_id})")
                    else:
                        print(f"       → {neighbor_id} (ingen person-data)")
            
            # Sjekk om ID-en finnes i noen persons relasjoner
            print(f"     Referert til av:")
            for person in helena_data.personer:
                if node_id in person.foreldre:
                    print(f"       → {person.fullt_navn}: som forelder")
                if node_id in person.barn:
                    print(f"       → {person.fullt_navn}: som barn")
                if node_id in person.partnere:
                    print(f"       → {person.fullt_navn}: som partner")
            print()
    
    if ikke_i_graf:
        print(f"\n🏝️ Personer uten forbindelser (isolerte noder):")
        for person_id in ikke_i_graf:
            person = next((p for p in helena_data.personer if p.id == person_id), None)
            if person:
                print(f"   - {person.fullt_navn} ({person_id})")
    
    print("\n" + "=" * 70)
    print("💡 Forklaring:")
    print("=" * 70)
    print("""
Manglende personer oppstår når:
1. En person er nevnt som forelder/barn/partner i GEDCOM-filen
2. Men denne personen selv mangler en INDI (Individual) record
3. Grafen legger til noden basert på relasjoner, men ingen Person-objekt eksisterer

Dette er vanlig når:
- GEDCOM-filen er ufullstendig
- Personer er referert til men ikke dokumentert
- Data er eksportert med filtre (f.eks. bare levende personer)
- Det er feil i GEDCOM-filen

Løsning:
- Sjekk MacFamilyTree og sikre at alle refererte personer har data
- Eksporter GEDCOM-filen på nytt med alle personer inkludert
- Eller oppdater import-koden til å håndtere manglende personer bedre
    """)
    
else:
    print("⚠️ Data ikke lastet inn ennå.")
````
